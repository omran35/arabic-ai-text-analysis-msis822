# المرحلة الرابعة: بناء النماذج وتقييمها


**الحالة: خطة فقط، لم تنفذ بعد.**

أبدأ بعد إنهاء الخصائص وتثبيت سياسة التقسيم.

1. أدرب خط أساس بسيط مثل Logistic Regression أو Naive Bayes.
2. أجرب مصنفين تقليديين إضافيين وأضبط الإعدادات باستخدام مجموعة التحقق.
3. أنفذ مسار التعلم العميق المحدد في التعليمات: شبكة على BERT embeddings أو fine-tuning.
4. أقيم النسخ النهائية على الاختبار بعد تثبيت الاختيارات.
5. أسجل Accuracy وPrecision وRecall وF1 وROC-AUC ومصفوفات الالتباس وتحليل الأخطاء.

**المدخلات المخططة:** الخصائص المحفوظة وتقسيم البيانات الثابت.
**المخرجات المخططة:** نماذج مدربة في models/، وجداول تقييم ورسوم وتحليل أمثلة الخطأ.

**ملاحظة:** البيانات غير متوازنة، لذلك لا أعتمد على Accuracy وحدها.
لا توجد نماذج مدربة أو نتائج تقييم في هذا الملف الآن.
